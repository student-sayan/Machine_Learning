```python
import pandas as pd
```

```python
customer_profile = pd.DataFrame(
    {
        "Customer_ID": [101, 102, 103, 104],
        "Name": ["Rahul Roy", "Priya Sen", "Amit Das", "Sneha Pal"],
        "Age": [28, 34, 22, 45],
        "City": ["Kolkata", "Mumbai", "Delhi", "Bangalore"]
    }
)

transaction_history = pd.DataFrame(
    {
        "Customer_ID": [101, 102, 101, 103, 102],
        "Transaction_ID": ["TXN1001", "TXN1002", "TXN1003", "TXN1004", "TXN1005"],
        "Amount": [1500, 2300, 450, 8900, 1200],
        "Date": [
            "2026-08-01",
            "2026-08-02",
            "2026-08-05",
            "2026-08-10",
            "2026-08-12",
        ]
    }
)
```

```python
df_merge = pd.merge(customer_profile,transaction_history, on="Customer_ID", how="inner")
print(df_merge)
```

```python
df_concat = pd.concat([customer_profile,transaction_history], axis=0, ignore_index=True)
print(df_concat)
```

```python
df_concat.to_csv("Project_1.csv")
```
