```python
import pandas as pd
```

```python
from textwrap import indent
data = {
    "Name" : ["Roy", "Sam", "Arun"],
    "Age" : [25,33,29],
    "Salary" : [30000,45000,35000]
}

df = pd.DataFrame(data)
print(df)
```

```python
#single Column
df.sort_values(by="Age", ascending=True, inplace=True)
print(df)
```

```python
#Multy Columns
df.sort_values(by=["Name","Age"], ascending=[True,False], inplace=True)
print(df)
```
