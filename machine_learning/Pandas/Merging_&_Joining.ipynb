```python
import pandas as pd
```

```python
df_customers = pd.DataFrame(
    {
        "Castomar_id" : [1,2,3],
        "Name" : ["Jos", "Adi", "Saka"]
    }
)

df_orders = pd.DataFrame(
    {
        "Castomar_id" : [1,2,4],
        "Order Amount" : [250,450,350]
    }
)

df_merge = pd.merge(df_customers, df_orders, on="Castomar_id", how="right") #how = inner, outer, left, right
print(df_merge)
```

```python
                                                         #========== Concatenatinate data ==========#

df_customers_1 = pd.DataFrame(
    {
        "Castomar_id" : [1,2,3],
        "Name" : ["Jos", "Adi", "Saka"]
    }
)

df_customers_2 = pd.DataFrame(
    {
        "Castomar_id" : [4,5,6],
        "Name" : ["Akash","Gopal","Raju"]
    }
)

df_concat = pd.concat([df_customers_1,df_customers_2], axis=0, ignore_index=True)
print(df_concat)
```

```python
df_concat = pd.concat([df_customers_1,df_customers_2], axis=1, ignore_index=True)
print(df_concat)
```
