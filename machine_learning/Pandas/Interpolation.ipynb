```python
import pandas as pd
```

```python
data = {
    "Name" : ["Ram","Shyam","Ghanshyam","Dhanshyam","Aditi","Jagdish","Raj","Simran"],
    "Age" : [28,None,22,30,29,40,25,32],
    "Salary" : [50000,None,45000,52000,49000,70000,48000,58000],
    "Performance Score" : [85,None,78,92,88,95,80,89]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Linear interpolation

#Formula = y=y1​+{(​x−x1)/(x2​−x1)​​}(y2​−y1​)

df["Age"] = df["Age"].interpolate(method="linear")
print(df)
```

```python
df["Salary"] = df["Salary"].interpolate(method="polynomial", order=2)
print(df)
```
