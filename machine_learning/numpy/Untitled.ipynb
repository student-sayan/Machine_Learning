```python
import numpy as np
```

```python
n = int(input("Enter the number of vectors = "))
result = np.array([0,0,0])
for i in range(n):
    print("\nEnter the vector", i+1)
    u = np.array(list(map(int,input().split())))
    a = int(input("Enter the scaler = "))
    result = result + a*u
    print(result)

if np.all(result == 0):
    print("Result is Zero Vector")
    print("Vectors may be Linearly Dependent")
else:
    print("Result is Not Zero Vector")
    print("Vectors may be Linearly Independent")
```

```python
A = np.array([[1,1,0],[0,1,1],[1,2,1]])
a_det = np.linalg.det(A)

if (a_det == 0):
    print("linearly dependent")
else:
    print("linearly independent")

k = np.linalg.matrix_rank(A)
print("Dimension = ",k)
```
