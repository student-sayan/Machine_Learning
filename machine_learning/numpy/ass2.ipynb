```python
import numpy as np
```

```python
mat = np.array([[5,6],
               [7,4]])
det1 = np.linalg.det(mat)
print(det1)

inv1 = np.linalg.inv(mat)
print(inv1)

t1 = np.transpose(mat)
print(t1)

print(mat*2)

adj1 = det1 * inv1
print(adj1)

rank1 = np.linalg.matrix_rank(mat)
print(rank1)

diag1 = np.diag(mat)
print(diag1)

trace1 = np.trace(mat)
print(trace1)

eigen_vector_values = np.linalg.eig(mat)
print(eigen_vector_values)
```
