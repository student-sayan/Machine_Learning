```python
import numpy as np
```

```python
list1 = np.array([10,20,30,40])
print(list1)
```

```python
print(list1+2)
print(list1-2)
print(list1*2)
print(list1/2)
```

```python
#aggregation function
print(np.sum(list1))
```

```python
print(np.min(list1))
print(np.max(list1))
print(np.std(list1))
print(np.var(list1))
```
