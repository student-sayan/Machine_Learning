```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
iris = load_iris()
x = iris.data

y = iris.target
target_names = iris.target_names
print(target_names)
```

```python
scaler = StandardScaler()
x_scaler = scaler.fit_transform(x)
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(x_scaler)
```

```python
colors = ['red','green','blue']


for i in range(3):
    plt.scatter(x_pca[y == i, 0],
                x_pca[y == i, 1],
                color=colors[i],
                )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Iris Dataset")

plt.show()
```

```python
#comulative Friquency
list1 = np.array([2,3,4,5,6])
list2 = np.array([8,12,15,10,5])

cf = np.cumsum(list2)
print(cf)
```

```python
list3 = np.array([2, 3, 4, 5])
f = np.array([3, 5, 4, 2])

mean = np.sum(list3*f)/np.sum(f)
print(f"mean = {mean:.2f}")
```

```python
cal = np.repeat(list3,f)
median = np.median(cal)
print(f"median = {median:.2f}")
```

```python
mode = list3[np.argmax(f)]
print(f"mode = {mode}")
```
