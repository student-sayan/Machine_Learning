```python
import numpy as np
```

```python
list1 = np.array([10,20,30,40,50])
new_list = np.insert(list1,2,60)
print(new_list)
```

```python
list_2d = np.array([[1,2],[3,4]])
newlist_2d = np.insert(list_2d,1,[5,6],axis=0)
print(newlist_2d)
```

```python
x = np.append(list1,[60,70,80])
print(x)
```

```python
#concatenate
a = np.array([10,20,30])
b = np.array([40,50,60])
c = np.concatenate((a,b),axis=0)
print(c)
```
