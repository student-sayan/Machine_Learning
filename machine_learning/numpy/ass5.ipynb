```python
import numpy as np
```

```python
#Gaussian Elimination

A = np.array([[1,1,1],
             [2,3,1],
             [1,2,2]])
B = np.array([6,10,9])

x = np.linalg.solve(A,B)
print(x)
```

```python
A = np.array([[1,1,1],
             [2,3,1],
             [1,2,2]])

B = np.array([6,10,9])

A_inv = np.linalg.inv(A) 

x = np.dot(A_inv,B)
print(x)

if np.linalg.matrix_rank(A) == np.linalg.matrix_rank(A|B):
    print("Consistent")
else:
    print("Inconsistent")
```
