```python
import numpy as np
```

```python
u = np.array([1,2,3])
v = np.array([4,-1,2])

print(u+v)
print(2*u)
dot_p = np.dot(u,v)
print(dot_p)

if dot_p == 0:
    print("orthogonal.")
else:
    print("Not")
```

```python
a1 = dot_p
mag_u = np.linalg.norm(u)
mag_v = np.linalg.norm(v)

meg = mag_u * mag_v

if a1 <= meg:
    print("Cauchy–Schwarz inequality")
else:
    print("Not")
```
