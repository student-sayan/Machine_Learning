```python
u = np.array([1,2,3])
v = np.array([4,5,6])
v_add = np.add(u,v)
print(v_add)
print(u*2)
dot_p = np.dot(u,v)
print(dot_p)
cross_p = np.cross(u,v)
print(cross_p)
if np.dot(u,v) == 0:
    print("Orthogonal")
else:
    print("not")

if np.all(np.cross(u,v)) == 0:
    print("Parallelism")
else:
    print("not")
```
