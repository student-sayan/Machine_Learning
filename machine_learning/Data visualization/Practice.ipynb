```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x1 = np.array([0,6])
y1 = np.array([0,250])
plt.plot(x1,y1,'o')
plt.show()
```

```python
x2 = np.array([1,2,6,8])
y2 = np.array([3,8,1,10])

plt.plot(x2,y2, 'o:g', ms=20, mfc='r')
plt.show()
```

```python
x3 = ['Mon','Tues','Wed','Thur','Fri']
y3 = [10,15,7,20,12]

plt.plot(x3,y3, color="black", linestyle='--', marker='o', linewidth=2.0, label="2026 sales date")
plt.title("Bekery sales this week")
plt.xlabel("Day of week")
plt.ylabel("Sales per day")
plt.legend()
plt.grid(color='green')
plt.savefig("Bekery.png", dpi=300, bbox_inches="tight")
plt.show()
```
