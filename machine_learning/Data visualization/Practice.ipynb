```python
import matplotlib.pyplot as plt
```

```python
l1 = [1,2,3,4,5,6]
l2 = [1000,2000,1500,3000,2500,4000]
plt.plot(l1,l2, linestyle='--',linewidth=2,marker='o',color='red',label = 'salary dts')
plt.xlabel("Months")
plt.ylabel("Amount earn")
plt.title("Amount graph")
plt.legend()
plt.grid()
plt.xticks([1,2,3,4,5,6],['m1','m2','m3','m4','m5','m6'])
plt.show()
```

```python
x = [1,2,3,4]
y = [10,20,15,25]

fig, ax = plt.subplots(1,2,figsize=(10,5))
ax[0].plot(x,y,color="green")
ax[0].set_title("Line plote")


ax[1].bar(x,y,color="purple")
ax[1].set_title("Bar plote")

fig.suptitle("Create subploting")
plt.tight_layout()
plt.savefig("sub1.png",dpi=300,bbox_inches="tight")
plt.show()
```
