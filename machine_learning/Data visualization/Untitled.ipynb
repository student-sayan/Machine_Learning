```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4]
y = [10,20,15,25]

plt.plot(x,y)
plt.show()
```

```python
x1 = ['Mon','Tues','Wed','Thur','Fri']
y1 = [10,15,7,20,12]
plt.plot(x1,y1)
plt.title("Bekery sales this week")
plt.xlabel("Day of the week")
plt.ylabel("Sales per day")
```

```python
months = [1,2,3,4]
sales = [1000,1500,1200,1800]
plt.plot(months, sales, color="blue", linestyle='--', marker='o', linewidth=2, label="2026 sales data")
plt.title("Monthly Sales Data Report")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.grid(color="red", linestyle=":", linewidth=1)
plt.xlim(1,4)
plt.ylim(0,2000)
plt.xticks([1,2,3,4],['m1','m2','m3','m4'])
plt.show()
```

```python
product = ['A','B','C','D']
sales = [1000,1500,800,1200]
plt.barh(product,sales,color="orange", label="sales of 2026")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales Comparison")
plt.legend()
plt.show()
```

```python
regions=['North','South','East','West']
ravenue=[3000,2000,1500,1000]
plt.pie(ravenue,labels=regions,autopct='%1.1f%%',colors=['lightgreen','blue','yellow','red'])
plt.title("Revenue Contribution By Region")
plt.legend()
plt.show()
```

```python
scores=[45,67,89,56,78,88,92,60,74,81,59,66,75,82,90,85,70,73,68,77]
plt.hist(scores,bins=5,color="yellow",edgecolor="orange")
plt.xlabel("Score Range")
plt.ylabel("Number of students")
plt.title("Distribution of students")
plt.show()
```

```python
hours_studies = [1,2,3,4,5,6,7,8]
exam_scors = [50,55,60,65,70,75,80,85]
plt.scatter(hours_studies,exam_scors, color = "green",marker="o",  label="Student data")
plt.xlabel("Hours studied")
plt.ylabel("Exam scors")
plt.title("Relationship Between study timre and exam scors")
plt.legend()
plt.grid()
plt.show()
```

```python
plt.scatter([1,2,3],[50,55,60], color="orange", label="Class A",marker="^")
plt.scatter([1,2,3],[46,50,52], color="blue", label="Class B")
plt.xlabel("Hours studied")
plt.ylabel("Exam scors")
plt.title("Relationship Between study timre and exam scors")
plt.legend()
plt.grid()
plt.show()
```

```python
Matches_Played = [2,5,7,1,12,15,18]
Goals_Scored = [1,4,5,2,7,12,11]
plt.scatter(Matches_Played,Goals_Scored,marker="o",color="green",label="2026 goals")
plt.xlabel("Matches Played")
plt.ylabel("Goals Scored")
plt.title("Season chart")
plt.grid()
plt.legend()
plt.show()
```
